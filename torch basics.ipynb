```python
import torch
import numpy as np
from torchvision import datasets, transforms
```

Tensor Creation

```python
x=torch.tensor([1,2,3])
y=torch.tensor([4,5,5])#Vector
a=torch.tensor([[1,2,3],[5,6,2],[7,8,9]])#Matrix
b=torch.tensor([[4,5,6],[5,90,23],[34,89,67]])#Matrix
p=torch.from_numpy(np.array([1,45,67]))#Numpy array to tensor
```

```python
print(a)
print(a.dtype)
```

Tensor Manipulation

1.Matrix multiplication

```python
c=torch.matmul(a,b)
print(c)
```

2.Vector addition

```python
z=torch.add(x,y)
print(z)
```

3.Element wise Multiplication

```python
w=torch.mul(x,y)
print(w)
```

4.Mean 

```python
mean=torch.mean(x.float())
print(mean)
```

5.Within Tensor Sum

```python
sum=torch.sum(p)
print(sum)
```

6.Concatenation

```python
m=torch.cat((x,y),dim=0)
m1=torch.sum(m)
print(m1)
```

7.RESHAPING

```python
print(a)
d=a.view(9)
print(d)
```

8.Element wise functions

```python
a1=torch.abs(d)
print(a1)

b1=torch.sqrt(c.float())
print(b1)
```

10.Some other functions

```python
max=torch.max(a)
print(max)
argmax=torch.argmax(c)
print(argmax)
```

Automatic differentiation

```python
x=torch.tensor([1.3,2.9,6.7],requires_grad=True)
y=x**3+4*x**2+5
y.sum().backward()
print(x.grad)
```

2.Custom dataset 

```python
from torch.utils.data import Dataset
class Customdataset(Dataset):
    def __init__ (self, data, labels, transform=None):
        self.data=data
        self.labels=labels
        self.transform=transform
    def __len__(self):
        return len(self.data)
    def __getitem__(self, idx):
        sample=self.data[idx]
        label=self.labels[idx]
        if self.transform:
            sample=self.transform(sample)
            return sample, label

train_samples=''''''
train_labels=''' '''   

#Initialize the custom dataset
customdataset=Customdataset(train_samples, train_labels, transform=transforms.ToTensor())        
```
